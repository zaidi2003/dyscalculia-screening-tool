# 07 - Threshold Matrix Calculation
This notebook calculates the threshold matrix for response time percentiles by grades for each question.

```python
import pandas as pd
```

```python
raw_data = pd.read_csv("../Data/raw.csv")
cols = list(raw_data)
cols
```

```python
raw_time_cols = [col for col in raw_data.columns if col.endswith('_Time') 
                 and 'by Grade' not in col 
                 and 'Score' not in col]
print(raw_time_cols)
```

```python
def calculate_thresholds(df, time_col, grade):
    grade_data = df[df['Grade'] == grade][time_col].dropna()
    return {
        'p10': grade_data.quantile(0.10),
        'p30': grade_data.quantile(0.30),
        'p70': grade_data.quantile(0.70),
        'p90': grade_data.quantile(0.90)
    }
```

```python
grades = [1, 2, 3]
threshold_matrix = []

for col in raw_time_cols:
    row = {'Question': col.replace('_Time', '')}
    for grade in grades:
        thresholds = calculate_thresholds(raw_data, col, grade)
        row[f'Grade{grade}_p10'] = thresholds['p10']
        row[f'Grade{grade}_p30'] = thresholds['p30']
        row[f'Grade{grade}_p70'] = thresholds['p70']
        row[f'Grade{grade}_p90'] = thresholds['p90']
    threshold_matrix.append(row)

threshold_df = pd.DataFrame(threshold_matrix)
threshold_df
```

```python
threshold_df.to_csv('../Data/threshold_matrix.csv', index=False)
print("Threshold matrix saved — shape:", threshold_df.shape)
```
